```python
import spacy

nlp = spacy.load("pt_core_news_sm")
texto = "O Professor Tiago explicou PLN usando o spaCy na Fatec."

doc = nlp(texto)

for token in doc:
    print(
        f"Texto: {token.text:10} | "
        f"Lemma: {token.lemma_:10} | "
        f"POS: {token.pos_:10} | "
        f"Stop: {token.is_stop}"
    )

print("\nEntidades:")
for ent in doc.ents:
    print(ent.text, ent.label_)
```

## Exercícios

### 1 - EXPLORANDO TOKENS

Dado o texto:

– "A inteligência artificial está transformando a educação superior no Brasil."

• Faça um programa que:

a) Mostre todos os tokens

b) Mostre apenas tokens que são palavras (ignore pontuação)

c) Conte o número total de tokens

```python
import spacy

nlp = spacy.load("pt_core_news_lg")
texto = "A inteligência artificial está transformando a educação superior no Brasil."

doc = nlp(texto)

print("\nTodos os tokens:")
for token in doc:
    print(
        f"Texto: {token.text}"
    )

print("\nTokens sem pontuação:")
for token in doc:
    if not token.is_punct:
        print(f"Texto: {token.text}")

print("\nNúmero total de tokens:", len(doc))
```

### 2 - LEMATIZAÇÃO E NORMALIZAÇÃO

Dado o texto:

– "Os alunos estavam estudando algoritmos e implementaram soluções
eficientes."

• Faça um programa que:

a) Mostre token.text e token.lemma_

b) Gere uma lista contendo apenas os lemas

c) Compare o texto original com o texto lematizado

```python
import spacy

nlp = spacy.load("pt_core_news_lg")
texto = "Os alunos estavam estudando algoritmos e implementaram soluções eficientes."

doc = nlp(texto)

for token in doc:
    print(
        f"Texto: {token.text}, "
        f"Lemma: {token.lemma_}, "
    )

print("\nLista de lemas:")
lemmas = [token.lemma_ for token in doc]
print(lemmas)
```

### 4 – STOPWORDS E ESTATÍSTICAS

Dado um parágrafo maior:

– Remova stopwords

– Remova pontuação

– Gere uma lista de palavras "limpas"

– Mostre as 5 palavras mais frequentes

• Objetivo:

– Introduzir pré-processamento para Machine Learning

```python
from collections import Counter

texto = """A inteligência artificial está transformando a educação superior no Brasil, permitindo novas metodologias de ensino e aprendizado. Os algoritmos de inteligência artifical são massa."""
doc = nlp(texto)

palavras_limpas = [
    token.text.lower()
    for token in doc
    if token.is_alpha and not token.is_stop
]

print("Palabras limpas:", palavras_limpas)

frequencia = Counter(palavras_limpas)
print("\nTop 5 palavras mais frequentes:")
print(frequencia.most_common(5))
```
