### Exercicio 4

Considere o seguinte conjunto de documentos:
- doc1 = `machine learning é usado para previsão de vendas`
- doc2 = `redes neurais são usadas em reconhecimento de imagem`
- doc3 = `análise de dados ajuda empresas a tomar decisões`
- doc4 = `algoritmos de aprendizado de máquina analisam grandes volumes de dados`

Agora considere a consulta do usuário:
- query = `aprendizado de máquina e dados`

Tarefas:
- Treine um vetor TF-IDF com os documentos.
- Transforme a consulta em vetor TF-IDF.
- Calcule a similaridade entre a consulta e cada documento.
- Exiba os documentos ordenados do mais relevante para o menos relevante.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
docs = [
    "machine learning é usado para previsão de vendas",
    "redes neurais são usadas em reconhecimento de imagem",
    "análise de dados ajuda empresas a tomar decisões",
    "algoritmos de aprendizado de máquina analisam grandes volumes de dados"
]

query = ["aprendizado de máquina e dados"]

vectorizer = TfidfVectorizer()

tfids_docs = vectorizer.fit_transform(docs)

tfids_query = vectorizer.transform(query)

similaridades = cosine_similarity(tfids_query, tfids_docs)

print("Similarides: ", similaridades)

indices = np.argsort(similaridades[0])[::-1]

print("\nDocumentos ordenados por relevância:")
for i in indices:
    print(docs[i])

```
