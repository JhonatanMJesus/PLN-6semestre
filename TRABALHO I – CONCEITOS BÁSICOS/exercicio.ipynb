```python
import spacy

nlp = spacy.load("pt_core_news_lg")

texto = "Após proibição, mais de 500 novas bets ilegais foram criadas, aponta levantamento. Maior volume foi registrado no dia seguinte à proibição, quando 292 plataformas foram criadas, segundo a Bet Legal. Em 2026, governo já encaminhou mais de 57 mil endereços para bloqueio. Por Isabela Ortiz, g1 — São Paulo 28/09/2026 17h30  Atualizado há 4 horas Um levantamento da plataforma Bet Legal mostra que 543 casas de apostas foram lançadas ilegalmente após a proibição das bets pelo governo federal na última sexta-feira (25). O número considera as novas casas identificadas entre as 18h de sexta-feira e a tarde desta segunda-feira (28). Apenas uma hora depois do anúncio da Medida Provisória, foram registradas 41 novas casas de apostas. O maior volume veio no dia seguinte, sábado (26), com 292 novas aberturas. O que acontece agora? Antes da proibição, as casas de apostas esportivas autorizadas pelo governo constavam em listas oficiais. Agora, elas têm até a próxima segunda-feira (5) para devolver o dinheiro dos apostadores que ainda tinham recursos nas plataformas na sexta-feira (25), último dia em que era possível apostar. 'A partir da publicação dessa medida provisória, as contas onde as plataformas de apostas recebem os aportes dos apostadores não serão mais admitidos nenhum novo aporte a partir da publicação da medida provisória', prosseguiu Dario Durigan, ministro da Fazenda. Mais de 57 mil endereços enviados para bloqueio em 2026. Os dados também apontam que 57.691 endereços — entre domínios e URLs — foram encaminhados pela Secretaria de Prêmios e Apostas à Anatel para bloqueio entre 1º de janeiro e 22 de setembro de 2026."

doc = nlp(texto)
```

```python
for token in doc:
    print(f"Token - {token.text}, Lemma - {token.lemma_} POS: {token.pos_}")
```

```python
for ent in doc.ents:
    print(f"Texto: {ent.text} - Tipo: {ent.label_}")
```
