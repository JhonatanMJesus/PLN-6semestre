## Classificador por Voz com Whisper

O whisper é um sistema avançado de reconhecimento automático de fala (ASR - Automatic Speech Recognition) desenvolvido e disponibilizado como código aberto pela Open AI. É mais amplamente conhecido como um poderoso transcritor e tradutor de voz.

```python
import sounddevice as sd
from scipy.io.wavfile import write
import whisper
from datetime import datetime
```

### Função para gravar áudio

```python
def gravar_audio(nome_arquivo="audio.wav", duracao=5, fs=44100):
        print("Gravando áudio...")
        audio = sd.rec(int(duracao * fs), samplerate=fs, channels=1)
        sd.wait()
        write(nome_arquivo, fs, audio)
        print("Gravação finalizada!")
```

### Carregando o modelo Whisper

```python
print('Carregando o modelo...')
model = whisper.load_model("large")
print("Modelo carregado")
```

### Função para transcrição

```python
def transcrever_audio(nome_arquivo="audio.wav"):
        print("Transcrevendo áudio...")
        result = model.transcribe(nome_arquivo, language="pt")
        texto = result["text"]
        print("Texto:", texto)
        return texto.lower()
```

### Classificador de comandos

```python
def classificar_comando(texto):
        if "luz" in texto and "ligar" in texto:
                print("Ligando a luz")
        elif "luz" in texto and "desligar" in texto:
                print("Desligar a luz")
        elif "horas" in texto:
                print("São ", datetime.now().strftime("%H:%M"))
        elif "data" in texto:
                print("Hoje é ", datetime.now().strftime("%d/%m/%Y"))
        else:
                print("Comando não reconhecido")
```

```python
texto_do_audio = ""

def executar_pipeline():
        gravar_audio()
        texto = transcrever_audio()
        classificar_comando(texto)
        global texto_do_audio
        texto_do_audio = texto

executar_pipeline()
```
